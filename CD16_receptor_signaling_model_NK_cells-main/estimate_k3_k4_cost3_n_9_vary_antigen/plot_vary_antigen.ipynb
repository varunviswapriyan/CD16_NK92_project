```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
lig = [60, 120, 180, 240, 300, 360, 420, 480, 500, 540, 600]

max_zeta = []
max_mixed = []
max_gamma = []

for _ in lig:
    data_zeta = np.loadtxt(f'zeta_results_{_}.txt', delimiter='\t')
    data_mixed = np.loadtxt(f'mixed_results_{_}.txt', delimiter='\t')
    data_gamma = np.loadtxt(f'gamma_results_{_}.txt', delimiter='\t')

    # Take max of second column (index 1)
    max_zeta.append(np.max(data_zeta[:, 1]))
    max_mixed.append(np.max(data_mixed[:, 1]))
    max_gamma.append(np.max(data_gamma[:, 1]))

# Bar plot with Arial font, size 24, grid off
import numpy as np
bar_width = 0.25
x = np.arange(len(lig))
size_font=16



colors=['#1f77b4','#87ceeb', '#ff7f0e']
plt.figure(figsize=(10, 6))
plt.bar(x - bar_width, max_zeta, width=bar_width, label='Zeta', color=colors[0])
plt.bar(x, max_mixed, width=bar_width, label='Mixed', color=colors[1])
plt.bar(x + bar_width, max_gamma, width=bar_width, label='Gamma', color=colors[2])
plt.xlabel('Ligand Concentration', fontname='Arial', fontsize=size_font)
plt.ylabel('Max of bound pZAP70 (uM)', fontname='Arial', fontsize=size_font)
#plt.title('Max of Second Column vs Ligand Concentration (Bar Plot)', fontname='Arial', fontsize=size_font)
plt.xticks(x, lig, fontname='Arial', fontsize=size_font)
plt.yticks(fontname='Arial', fontsize=size_font)
plt.legend(fontsize=size_font)
plt.grid(False)
plt.tight_layout()
plt.show()
```

```python
# Dot plot for max of bound pZAP70 (uM) vs ligand concentration
plt.figure(figsize=(10, 6))
plt.plot(lig, max_zeta, 'o-', label='Zeta', markersize=10, color=colors[0])
plt.plot(lig, max_mixed, 'o-', label='Mixed', markersize=10, color=colors[1])
plt.plot(lig, max_gamma, 'o-', label='Gamma', markersize=10, color=colors[2])
plt.xlabel('Ligand Concentration', fontname='Arial', fontsize=size_font)
plt.ylabel('Max of bound pZAP70 (uM)', fontname='Arial', fontsize=size_font)
#plt.title('Max of Second Column vs Ligand Concentration (Dot Plot)', fontname='Arial', fontsize=size_font)
plt.xticks(fontname='Arial', fontsize=size_font)
plt.yticks(fontname='Arial', fontsize=size_font)
plt.legend(fontsize=size_font)
plt.grid(False)
plt.tight_layout()
plt.show()
```
